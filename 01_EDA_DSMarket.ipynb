## 2. Paleta de marca DSMarket

Antes de graficar nada, definimos una paleta propia en vez de usar los colores por defecto de matplotlib — para que todos los gráficos de este proyecto (EDA, clustering, dashboard) se vean como parte de un mismo documento de DSMarket.

Saqué los colores directamente del logo de la empresa (extraje la imagen del enunciado y muestreé los píxeles): un azul marino `#0a2a5c`, un verde lima `#81b80c`, y un dorado `#fabc17`. Como pediste que la paleta esté más centrada en el azul, lo uso como color principal (títulos, gráficos de una sola serie, y la escala secuencial para magnitudes), y dejo el verde y el dorado como colores de apoyo para cuando necesitamos distinguir categorías (ciudades, comparaciones).

No elegí los tonos "a ojo": para el uso en gráficos (texto/marcas sobre fondo blanco) probé varias variantes y validé cada combinación con un script que chequea que los colores se puedan distinguir con distintos tipos de daltonismo (protanopía/deuteranopía/tritanopía) y que tengan contraste suficiente — casi el mismo azul y dorado del logo ya pasaban la validación tal cual; para el verde tuve que ajustar el tono para que no se confundiera con el dorado en gráficos con las tres ciudades juntas.


```python
# Paleta DSMarket — derivada del logo de la empresa, ajustada y validada para gráficos
# (contraste vs. fondo blanco + distinción bajo daltonismo protan/deutan/tritan)

DSMARKET_NAVY = "#0a2a5c"     # azul oscuro del logo -> texto, títulos, énfasis (no para marcas de datos: muy oscuro)
DSMARKET_BLUE = "#2a78d6"     # azul de marca -> color principal: series únicas, ranking, escala secuencial
DSMARKET_GOLD = "#eda100"     # dorado del logo -> color de apoyo 1
DSMARKET_GREEN = "#6a9f1f"    # verde lima del logo, reajustado en luminosidad para distinguirse del dorado

# Paleta categórica para comparar las 3 ciudades piloto (orden fijo: nunca reasignar estos colores)
CITY_COLORS = {
    "Boston": DSMARKET_GREEN,
    "New York": DSMARKET_BLUE,
    "Philadelphia": DSMARKET_GOLD,
}

# Rampa secuencial de un solo tono (azul de marca, claro -> oscuro) para magnitudes / rankings de una sola serie
DSMARKET_BLUE_RAMP = ["#cde2fb", "#9ec5f4", "#6da7ec", "#3987e5", "#2a78d6", "#1c5cab", "#0d366b"]
dsmarket_blues_cmap = LinearSegmentedColormap.from_list("dsmarket_blues", DSMARKET_BLUE_RAMP)

sns.set_theme(style="whitegrid", rc={
    "axes.edgecolor": "#c3c2b7",
    "axes.labelcolor": "#0b0b0b",
    "text.color": "#0b0b0b",
    "xtick.color": "#52514e",
    "ytick.color": "#52514e",
    "grid.color": "#e1e0d9",
    "figure.facecolor": "white",
    "axes.facecolor": "white",
})
plt.rcParams["axes.titlecolor"] = DSMARKET_NAVY
plt.rcParams["axes.titleweight"] = "bold"

# swatch rápido para verlos
fig, ax = plt.subplots(figsize=(6, 1))
swatches = [DSMARKET_NAVY, DSMARKET_BLUE, DSMARKET_GREEN, DSMARKET_GOLD]
labels = ["navy (texto)", "blue (marca)", "green", "gold"]
for i, (c, l) in enumerate(zip(swatches, labels)):
    ax.add_patch(plt.Rectangle((i, 0), 1, 1, color=c))
    ax.text(i + 0.5, -0.3, l, ha="center", va="top", fontsize=9)
ax.set_xlim(0, len(swatches))
ax.set_ylim(-0.6, 1)
ax.axis("off")
plt.show()
```

**Validación (script de la skill de dataviz, chequeo de daltonismo y contraste):**

- `blue + green + gold` (orden Boston→New York→Philadelphia, el que uso en los gráficos de ciudad): **todos los checks OK**, sin advertencias — es el orden que evita que el verde y el dorado queden pegados uno al lado del otro.
- `blue + gold` solo (para comparar día de evento vs. día normal más adelante): **todos los checks OK**.
- Los 3 colores tienen buen contraste salvo el dorado, que por ser un tono claro queda algo bajo de contraste sobre fondo blanco — por eso en los gráficos donde aparece siempre va acompañado de su etiqueta directa (leyenda o texto), nunca solo el color.

### 4.3. Prices

**Duda que surgió acá:** `item_prices.csv` tiene **243.920 filas (3.5%) sin `yearweek`**. `item` y `store_code` sí están completos en esas filas (o sea, en principio sabemos a qué producto-tienda corresponden), pero sin semana no sabemos en qué momento del tiempo ubicar el precio. Investigamos el patrón antes de decidir qué hacer.

```python
null_rows = prices[prices['yearweek'].isna()]

# ¿Están concentradas en alguna tienda/categoría?
print("Filas nulas por store_code:")
print(null_rows['store_code'].value_counts())
print()
print("Items distintos en filas nulas:", null_rows['item'].nunique(), "de", prices['item'].nunique(), "totales")
print()

# ¿Cuántas filas nulas hay por combinación item+store_code?
nulls_per_group = null_rows.groupby(['item', 'store_code']).size()
print("Filas nulas por grupo (item, store_code):")
print(nulls_per_group.describe())
print()

# ¿Cuál es la última yearweek válida (no nula) de cada grupo?
last_valid = prices.dropna(subset=['yearweek']).groupby(['item', 'store_code'])['yearweek'].max()
print("Última yearweek válida por grupo — ¿es siempre la misma?")
print(last_valid.value_counts())

# ¿Las filas nulas están siempre al final de cada grupo (orden del archivo), o intercaladas?
def nulls_only_at_tail(df):
    is_null = df['yearweek'].isna().values
    if not is_null.any():
        return True
    return is_null[is_null.argmax():].all()

tail_check = prices.groupby(['item', 'store_code'], sort=False).apply(nulls_only_at_tail)
print()
print("¿En todos los grupos los nulos están solo al final?:", tail_check.all())
```

**Hallazgo:** no es un problema aleatorio — es un patrón perfectamente sistemático. **Las 243.920 filas nulas son, en todos y cada uno de los 30.490 grupos (item, tienda), exactamente las últimas 8 filas** (en orden del archivo), y en todos los grupos la última `yearweek` válida es **201617**. Es decir: son 8 semanas de precio que existen (con su `sell_price`) pero perdieron la etiqueta de semana, y son siempre las 8 semanas *posteriores* a la 201617.

Y acá está la clave: el calendario de ventas termina el **2016-04-24**, que cae justo alrededor de esa misma semana 201617. O sea, **estas 8 semanas "sin nombre" quedan fuera del rango de fechas que cubre `item_sales.csv`** — no hay ningún día de venta con el que cruzarlas de todos modos.

**Conclusión:** las descartamos con confianza (no es una pérdida de información real para este proyecto, porque no hay ventas en ese rango de fechas para cruzarlas). Si en algún momento se sumaran más semanas de ventas al dataset, se podrían recuperar fácilmente incrementando secuencialmente desde 201617 (ya que confirmamos que están siempre ordenadas y son siempre las mismas 8 semanas por grupo) — lo dejamos anotado por si hace falta más adelante.

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
for ax, city in zip(axes, CITY_COLORS):
    top10 = sales_by_region_item[city].sort_values(ascending=False).head(10).sort_values()
    ax.barh(top10.index, top10.values, color=CITY_COLORS[city])
    ax.set_title(city, color=DSMARKET_NAVY)
    ax.set_xlabel('Unidades')
fig.suptitle('Top 10 productos por ciudad', color=DSMARKET_NAVY, fontweight='bold', y=1.04)
plt.tight_layout()
plt.show()
```

Para visualizar la heterogeneidad entre tiendas de una forma más completa que el conteo de "top 15 en común", calculamos qué tan parecido es el catálogo más vendido entre cada par de las 10 tiendas (similitud de Jaccard sobre el top 50 de productos de cada una) y lo miramos como mapa de calor. Esto también nos sirve de anticipo para la Tarea 2 (clustering de tiendas) que pidió Joelle.

```python
# Matriz tienda x producto (unidades totales), sin melt: sumamos directo sobre el formato ancho
store_item_totals = sales_clean.groupby(['store_code', 'item'])[day_cols].sum().sum(axis=1).unstack('item').fillna(0)

def top_set(row, n=50):
    return set(row.sort_values(ascending=False).head(n).index)

top_sets = {store: top_set(store_item_totals.loc[store]) for store in store_item_totals.index}
stores_order = sorted(store_item_totals.index, key=lambda s: (s.split('_')[0], s))  # agrupa por ciudad

jaccard = pd.DataFrame(index=stores_order, columns=stores_order, dtype=float)
for a in stores_order:
    for b in stores_order:
        inter = len(top_sets[a] & top_sets[b])
        union = len(top_sets[a] | top_sets[b])
        jaccard.loc[a, b] = inter / union

fig, ax = plt.subplots(figsize=(8, 7))
im = ax.imshow(jaccard.values.astype(float), cmap=dsmarket_blues_cmap, vmin=0, vmax=1)
ax.set_xticks(range(len(stores_order)))
ax.set_xticklabels(stores_order, rotation=45, ha='right')
ax.set_yticks(range(len(stores_order)))
ax.set_yticklabels(stores_order)
for i in range(len(stores_order)):
    for j in range(len(stores_order)):
        v = jaccard.values[i, j]
        ax.text(j, i, f"{v:.2f}", ha='center', va='center', fontsize=8,
                color='white' if v > 0.55 else DSMARKET_NAVY)
ax.set_title('Similitud entre tiendas (Jaccard, top-50 productos)', color=DSMARKET_NAVY, fontweight='bold')
fig.colorbar(im, ax=ax, label='Similitud (0 = nada en común, 1 = idéntico)')
plt.tight_layout()
plt.show()
```

**Hallazgo:** se ve con claridad que las tiendas de una misma ciudad son más parecidas entre sí que con tiendas de otra ciudad (bloques más claros/oscuros agrupados en la diagonal), y dentro de eso, algunas tiendas de Filadelfia se despegan del resto — consistente con lo que vimos arriba. Este mapa de similitud es prácticamente un primer boceto de la Tarea 2 (clustering de tiendas): ya se insinúan grupos naturales antes de correr ningún algoritmo.

## 10. Comparativa de precios entre tiendas

Michelle: *"nuestras acciones de marketing tendrían que explotar esas diferencias"* — miramos si un mismo producto se vende a precios distintos según la tienda.

```python
avg_price_by_item_store = (prices_clean.groupby(['item', 'store_code'])['sell_price']
                            .mean().unstack('store_code'))

# Para cada producto vendido en más de una tienda, medimos la dispersión de precio entre tiendas
price_dispersion = avg_price_by_item_store.agg(['min', 'max', 'std'], axis=1)
price_dispersion['range_pct'] = (price_dispersion['max'] - price_dispersion['min']) / price_dispersion['min'] * 100
price_dispersion = price_dispersion.dropna().sort_values('range_pct', ascending=False)

print(f"Productos con precio en más de una tienda: {len(price_dispersion)}")
print(f"Diferencia de precio entre tienda más cara y más barata (% sobre la más barata):")
print(price_dispersion['range_pct'].describe())
print()
print("Los 10 productos con MAYOR diferencia de precio entre tiendas:")
price_dispersion.head(10)
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(price_dispersion['range_pct'].clip(upper=100), bins=40, color=DSMARKET_BLUE)
ax.axvline(price_dispersion['range_pct'].median(), color=DSMARKET_GOLD, linestyle='--', linewidth=2,
           label=f"mediana: {price_dispersion['range_pct'].median():.1f}%")
ax.set_xlabel('Diferencia de precio entre tienda más cara y más barata (%, recortado a 100%)')
ax.set_ylabel('Cantidad de productos')
ax.set_title('¿Cuánto varía el precio de un mismo producto entre tiendas?')
ax.legend(frameon=False)
plt.tight_layout()
plt.show()
```

Sigamos las dos preguntas que quedaron abiertas: ¿es un patrón geográfico (una ciudad sistemáticamente más cara) o depende más de la categoría de producto?

```python
item_category_map = sales_clean[['item', 'category']].drop_duplicates().set_index('item')['category']
price_dispersion_cat = price_dispersion.copy()
price_dispersion_cat['category'] = price_dispersion_cat.index.map(item_category_map)

fig, ax = plt.subplots(figsize=(7, 4))
cats = sorted(price_dispersion_cat['category'].dropna().unique())
data = [price_dispersion_cat.loc[price_dispersion_cat['category'] == c, 'range_pct'].clip(upper=100) for c in cats]
bp = ax.boxplot(data, tick_labels=cats, showfliers=False, patch_artist=True, widths=0.5)
for patch in bp['boxes']:
    patch.set_facecolor(DSMARKET_BLUE)
    patch.set_alpha(0.85)
for median in bp['medians']:
    median.set_color(DSMARKET_GOLD)
    median.set_linewidth(2)
ax.set_ylabel('% dif. de precio entre tiendas (máx 100%)')
ax.set_title('Dispersión de precios entre tiendas, por categoría', color=DSMARKET_NAVY, fontweight='bold')
fig.subplots_adjust(left=0.16)
plt.tight_layout()
plt.show()

print(price_dispersion_cat.groupby('category')['range_pct'].median().sort_values(ascending=False))
```

Pero la **dispersión** de precios entre tiendas para un mismo producto sí varía por categoría, y con un resultado que no era el esperable: **`SUPERMARKET` es la categoría con MÁS dispersión de precio entre tiendas** (mediana 2.75%), por encima de `ACCESORIES` (1.18%) y `HOME_&_GARDEN` (0.97%, la más uniforme). ### 10.1. ¿Hay una tienda puntual que empuje la dispersión, o es un patrón de ciudad?

La vista anterior mide dispersión por producto (min/max entre tiendas) y por categoría, pero no dice si hay una tienda en particular que sea sistemáticamente más cara o más barata — relevante porque una acción de marketing se ejecuta tienda por tienda, no por categoría. Para aislar el efecto "tienda" del efecto "qué productos vende cada tienda", normalizamos el precio de cada producto contra el precio promedio de ese mismo producto en las tiendas donde se vende, y promediamos esa relación por tienda: un índice de 0% significa que la tienda cobra, en promedio, lo mismo que el resto por los mismos productos.

```python
item_mean_price = avg_price_by_item_store.mean(axis=1)
relative_price = avg_price_by_item_store.div(item_mean_price, axis=0)  # 1.0 = precio promedio del producto en sus tiendas

store_meta = sales_clean[['store_code', 'store', 'region']].drop_duplicates().set_index('store_code')

store_price_index = relative_price.mean(axis=0).to_frame('price_index')
store_price_index['price_index_pct'] = (store_price_index['price_index'] - 1) * 100
store_price_index = store_price_index.join(store_meta).sort_values(['region', 'price_index_pct'], ascending=[True, False])

print("Índice de precio por tienda (0% = cobra igual que el promedio del producto en sus tiendas):")
store_price_index[['store', 'region', 'price_index_pct']]
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
order = store_price_index.sort_values(['region', 'price_index_pct'])
colors = [CITY_COLORS[c] for c in order['region']]
ax.barh(order['store'], order['price_index_pct'], color=colors)
ax.axvline(0, color=DSMARKET_NAVY, linewidth=1)
ax.set_xlabel('Precio vs. promedio del producto en sus tiendas (%)')
ax.set_title('¿Qué tiendas cobran sistemáticamente más o menos?', color=DSMARKET_NAVY, fontweight='bold')
handles = [plt.Rectangle((0, 0), 1, 1, color=CITY_COLORS[c]) for c in CITY_COLORS]
ax.legend(handles, CITY_COLORS.keys(), frameon=False, loc='lower right')
plt.tight_layout()
plt.show()
```

```python
# Para cada tienda, ¿en qué % de sus productos cobra por encima del promedio de ese producto?
# (evita que un puñado de productos con diferencias grandes pese más de la cuenta en el promedio)
above_below = (
    (relative_price > 1.001).sum(axis=0).to_frame('n_mas_caro')
    .join((relative_price < 0.999).sum(axis=0).to_frame('n_mas_barato'))
)
above_below['n_total'] = relative_price.notna().sum(axis=0)
above_below['pct_mas_caro'] = (above_below['n_mas_caro'] / above_below['n_total'] * 100).round(1)
above_below = above_below.join(store_meta).sort_values(['region', 'pct_mas_caro'], ascending=[True, False])

print("% de productos en los que cada tienda cobra por encima del promedio de ese producto:")
above_below[['store', 'region', 'pct_mas_caro']]
```

**Hallazgo:** el efecto que domina es de **ciudad**, no de una tienda puntual — con una excepción clara en New York. Las 3 tiendas de **Boston** quedan sistemáticamente por debajo del promedio del producto (entre -0.85% y -0.55%, las 3 negativas: cobran de más en solo 20-30% de sus productos), y las 3 de **Philadelphia** sistemáticamente por encima (entre +0.32% y +0.67%, las 3 positivas: cobran de más en 41-55% de sus productos). En **New York** el patrón es distinto: Greenwich Village, Harlem y Brooklyn cobran de más en ~47-50% de sus productos, pero **Tribeca es la única tienda de NY que queda por debajo del promedio** (solo 25.9% de sus productos por encima, la mitad que sus 3 vecinas) — es la tienda puntual que "desentona" dentro de su ciudad, y explica buena parte de por qué NY es la ciudad con el spread interno más ancho (0.79 puntos porcentuales entre su tienda más cara y más barata, vs. ~0.30-0.35 en Boston y Filadelfia).

Importante para la lectura: **las magnitudes acá son chicas** (todas dentro de ±1%) frente a la dispersión por producto que vimos antes (mediana de 2.75% en SUPERMARKET, con productos puntuales llegando a decenas de %). Conclusión: la dispersión de precio entre tiendas para un mismo producto **no está explicada, en general, por una tienda que sea cara o barata en todo su catálogo** — es mayormente idiosincrática, producto por producto. El único patrón sistemático real es un leve efecto de ciudad (Boston algo más barato, Filadelfia algo más caro) y el caso puntual de Tribeca en NY. Para Michelle: si la acción de marketing busca "la tienda barata" como gancho, Tribeca es el caso más claro y accionable con una historia coherente detrás; para el resto, la variación real está a nivel producto (sección anterior), no a nivel tienda.

## 11. Estacionalidad y eventos especiales

El calendario original solo marca SuperBowl, Thanksgiving, NewYear, Easter y Ramadan. Sumamos dos eventos calculables por regla de fecha que sí tienen presencia constante en el negocio real: **Christmas** (25 de diciembre) y **Labor Day** (primer lunes de septiembre). También evaluamos agregar **Black Friday** (día siguiente a Thanksgiving), pero esas 5 fechas no muestran ningún patrón distinguible frente a un día normal, así que no lo tratamos como un evento importante del calendario extendido — sin un equipo de datos real al que consultarle cómo se generó el dataset, no tiene sentido dejarlo como hallazgo abierto. Dejamos afuera también "vuelta al cole" porque no es una fecha única sino un período de varias semanas — no encaja bien en un calendario de "un evento por día"; si querés, más adelante lo podemos modelar como ventana de fechas en vez de día puntual.

```python
cal_idx = calendar_clean.set_index('date')
event_days = cal_idx['event'].dropna()
cal_min, cal_max = calendar_clean['date'].min(), calendar_clean['date'].max()

extra_events = {}

# Christmas (25/12) y Labor Day (primer lunes de septiembre), para cada año cubierto por el calendario
for y in calendar_clean['date'].dt.year.unique():
    christmas = pd.Timestamp(year=int(y), month=12, day=25)
    if cal_min <= christmas <= cal_max:
        extra_events[christmas] = 'Christmas'

    labor_day = pd.Timestamp(year=int(y), month=9, day=1)
    while labor_day.weekday() != 0:  # 0 = lunes
        labor_day += pd.Timedelta(days=1)
    if cal_min <= labor_day <= cal_max:
        extra_events[labor_day] = 'Labor Day'

print(f"Eventos nuevos calculados: {len(extra_events)}")
for d, e in sorted(extra_events.items()):
    print(d.date(), e)

event_days_extended = pd.concat([event_days, pd.Series(extra_events)]).sort_index()
print()
print("Calendario de eventos extendido:")
print(event_days_extended.value_counts())
```

```python
daily_total = sales_by_region.sum(axis=1)  # ya calculado en la sección 7
daily_total.index = pd.to_datetime(daily_total.index)

avg_by_event = daily_total.groupby(daily_total.index.map(lambda d: event_days_extended.get(d, 'Día normal'))).mean()
avg_by_event = avg_by_event.sort_values()

fig, ax = plt.subplots(figsize=(8, 6))
colors = [DSMARKET_GOLD if idx != 'Día normal' else DSMARKET_BLUE for idx in avg_by_event.index]
ax.barh(avg_by_event.index, avg_by_event.values, color=colors)
ax.set_xlabel('Unidades vendidas promedio (todas las ciudades)')
ax.set_title('Ventas promedio: eventos especiales vs. día normal (calendario extendido)', color=DSMARKET_NAVY, fontweight='bold')
for i, v in enumerate(avg_by_event.values):
    ax.text(v, i, f' {v:,.0f}', va='center', fontsize=9)
plt.tight_layout()
plt.show()
```

**Hallazgo:** al sumar Christmas, se ve que las ventas ese día son prácticamente **cero (16 unidades)** — es decir, el negocio cierra directamente el 25 de diciembre. Eso confirma el patrón de "cierre en feriados grandes" que ya habíamos visto con Thanksgiving.

Los eventos que muestran un pico claro y consistente por encima del día normal son **Labor Day, SuperBowl y Easter** (subas del 15-22% sobre el día normal), y **Ramadan starts** apenas por encima del normal (+1%). Evaluamos Black Friday y no mostró ningún efecto distinguible en estos datos, así que no lo dejamos como evento destacado del calendario (ver nota de la sección anterior).

```python
weekday_map = calendar_clean.set_index('date')['weekday']
daily_total_df = daily_total.rename('units').to_frame()
daily_total_df['weekday'] = daily_total_df.index.map(weekday_map)
daily_total_df['month'] = daily_total_df.index.month

weekday_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekday_es = {'Monday': 'Lunes', 'Tuesday': 'Martes', 'Wednesday': 'Miércoles', 'Thursday': 'Jueves',
              'Friday': 'Viernes', 'Saturday': 'Sábado', 'Sunday': 'Domingo'}
avg_by_weekday = daily_total_df.groupby('weekday')['units'].mean().reindex(weekday_order)

month_names = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']
avg_by_month = daily_total_df.groupby('month')['units'].mean()

fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
axes[0].bar([weekday_es[d] for d in avg_by_weekday.index], avg_by_weekday.values, color=DSMARKET_BLUE)
axes[0].set_title('Ventas promedio por día de la semana', color=DSMARKET_NAVY, fontweight='bold')
axes[0].set_ylabel('Unidades')
axes[0].tick_params(axis='x', rotation=30)

axes[1].bar([month_names[m - 1] for m in avg_by_month.index], avg_by_month.values, color=DSMARKET_BLUE)
axes[1].set_title('Ventas promedio por mes del año', color=DSMARKET_NAVY, fontweight='bold')
axes[1].set_ylabel('Unidades')

plt.tight_layout()
plt.show()
```

**Hallazgo:** hay estacionalidad semanal clara (fin de semana por encima de días de semana, patrón típico de retail) y estacionalidad mensual también — vale la pena comentarlo en la memoria como parte del comportamiento estructural de la demanda, más allá de los eventos puntuales. Esto también es información directamente reutilizable como *features* de calendario para la Tarea 3 (forecasting).
